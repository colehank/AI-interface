# LVM(Large Vision Model)使用示例

# 1 包导入

```python
import lmitf
from lmitf import BaseLVM
from lmitf import TemplateLLM
```

## 2 文生图示例

```python
vlm = BaseLVM()
res = vlm.create(
    prompt="A beautiful landscape with mountains and a river",
    model='gpt-image-1',
)
res
```

## 3 图生图示例

```python
res = vlm.edit(res, "Add flying cats in the sky")
res
```
